```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("/content/OIP.jpg")

# Get image dimensions
rows, cols = image.shape[:2]

# Source points
src_pts = np.float32([
    [50, 50],
    [300, 50],
    [50, 300],
    [300, 300]
])

# Destination points
dst_pts = np.float32([
    [0, 0],
    [cols - 1, 0],
    [80, rows - 1],
    [cols - 80, rows - 1]
])

# Compute Homography matrix
H, status = cv2.findHomography(src_pts, dst_pts)

# Apply Homography transformation
result = cv2.warpPerspective(image, H, (cols, rows))

# Display images
cv2_imshow(image)
cv2_imshow(result)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the input image
image = cv2.imread("/content/OIP (1).jpg")

# Get image dimensions
rows, cols = image.shape[:2]

# Source points
src_points = np.float32([
    [50, 50],
    [300, 50],
    [50, 300],
    [300, 300]
])

# Destination points
dst_points = np.float32([
    [20, 80],
    [cols - 50, 30],
    [80, rows - 30],
    [cols - 20, rows - 50]
])

# Compute Homography matrix (DLT)
H, status = cv2.findHomography(src_points, dst_points)

# Apply perspective transformation
transformed = cv2.warpPerspective(image, H, (cols, rows))

# Display images
cv2_imshow(image)
cv2_imshow(transformed)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("/content/VK18.webp")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Sobel operator along X-axis
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)

# Apply Sobel operator along Y-axis
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

# Convert to absolute values
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine X and Y edges
sobel_xy = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

# Display images
cv2_imshow(image)
cv2_imshow(sobel_x)
cv2_imshow(sobel_y)
cv2_imshow(sobel_xy)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("/content/VK18.webp")

# Define Laplacian mask with negative center coefficient
kernel = np.array([[0, -1, 0],
                   [-1, 4, -1],
                   [0, -1, 0]])

# Apply the Laplacian filter
laplacian = cv2.filter2D(image, -1, kernel)

# Sharpen the image
sharpened = cv2.subtract(image, laplacian)

# Display images
cv2_imshow(image)
cv2_imshow(sharpened)
```
